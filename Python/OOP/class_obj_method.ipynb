```python
class computer():
    def __init__(self,i,r,d): #__init__ means initialize the varible
        self.i = i
        self.r = r
        self.d = d
        print("in init method")
        
    def config(self):
            print(self.i,self.r,self.d) 
            
        
        
com1=computer("10th gen","16gb","1TB") 
com2=computer("11th gen","16gb","1TB")

com1.config()
com2.config()
```

```python
class car():
    def __init__(self):
        self.mil = 10
        self.com = 'bmw'
        
c1 = car()
c2 = car()

print(c1.com, c1.mil)
print(c2.com, c2.mil)

print(id(c1))
print(id(c2))
```
