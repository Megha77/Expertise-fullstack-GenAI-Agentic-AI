## 47. Write a program to find the average of a list of numbers entered by the user.

```python
num1= input("Enter Number seperated by spaces:")
num_split=num1.split()

numbers=[]

for num in num_split:
    numbers.append(int(num))
    
total_sum=sum(numbers)
len_num=len(numbers)
avg=total_sum/len_num
print("Average of numbers is:", avg)
```

## 49. How do you check if a string entered by the user contains any punctuation?

```python
import string
text=input("Enter String:")
count=0

for char in text:
    if char in string.punctuation:
        #print("String has punctuation")
        count += 1
    else:
        count += 0
        
if count>0:
    print("String has punctuation")
else:
    print("String has no punctuation")
```

## 50. Write a program that accepts a sentence and prints the longest word.

```python
sentence=input("Enter String space separated:")

words=sentence.split()
word_count=len(words)
word_list_lengths=[]
longest_word=""
max_length=0

for w in words:
    len_word=len(w)
    word_list_lengths.append(len_word)
    if len_word>max_length:
        max_length=len_word
        longest_word=w
    
max_length=max(word_list_lengths)    

print("Number of words in the sentence:", word_count)
print("Maximum word length:", max_length)
print("Longest word in the sentence is:", longest_word)

```
