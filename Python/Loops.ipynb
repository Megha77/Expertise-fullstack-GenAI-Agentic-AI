```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam", "Headset"]
stock = [0, 5, 42, 3, 0, 17]

inventory_report = []

out_of_stock = 0
low = 0
ok = 0

for index, product in enumerate(products, start=1):
    qty = stock[index - 1]

    if qty == 0:
        status = "OUT OF STOCK"
        out_of_stock += 1
    elif 1 <= qty <= 10:
        status = "LOW"
        low += 1
    else:
        status = "OK"
        ok += 1

    inventory_report.append({
        "index": index,
        "product": product,
        "qty": qty,
        "status": status
    })

result = {
    "inventory_report": inventory_report,
    "summary": {
        "total_products": len(products),
        "out_of_stock": out_of_stock,
        "low": low,
        "ok": ok
    }
}

print(result)
```
