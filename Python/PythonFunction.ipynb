```python
a=10

def function1():
    global a
    a=a+15
    print('in function',a)

function1()
    
print('out function',a)
```

```python
a=10 # global variable

def function1():
    globals()['a']+=11 #global function
    print('in function',a)

function1()
    
print('out function',a)
```

```python
lst=[10,20,55,33,40,90,70,42]
def numFind(lst):
    even=0
    odd=0
    
    for i in lst:
        if i%2==0:
            even+=1
        else:
            odd+=1 
    return even,odd

even,odd=numFind(lst)

print("Even count:{} and Odd count:{}".format(even,odd))
```
