```python
from nltk.util import bigrams,trigrams,ngrams
```

```python
string='generate ai & agentic ai is automate the industry & more focus on every domain'
quotes_tokens=nltk.word_tokenize(string)
quotes_tokens
```

```python
quotes_bigram=list(nltk.bigrams(quotes_tokens))
quotes_bigram
```

```python
quotes_trigram=list(nltk.trigrams(quotes_tokens))
quotes_trigram
```

```python
quotes_ngrams=list(nltk.ngrams(quotes_tokens,6))
quotes_ngrams
```

```python
from nltk.stem import PorterStemmer
pst=PorterStemmer()
```

```python
pst.stem('affection')
```

```python
pst.stem('playing')
```

```python
pst.stem('maximum')
```

```python
pst.stem('sought')
```

```python
pst.stem('dictonary')
```

```python
words_to_stem=['give','giving','given','gaved','thinking','loving','maximum','megha']
for words in words_to_stem:
    print(words+ ' : ' + pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer
lst= LancasterStemmer()

for words in words_to_stem:
    print(words+ ' : ' + lst.stem(words))
```

```python
from nltk.stem import wordnet
from nltk.stem import WordNetLemmatizer
word_len= WordNetLemmatizer()
```

```python
words_to_stem
```

```python
for words in words_to_stem:
    print(words+ ': ' + word_len.lemmatize(words))
```
